# Etape 1 : Génération de donnnées synthétiques 

```python
# Chager les données les modules
import pandas as pd
import numpy as np
from sdv.single_table import CTGANSynthesizer
from sdv.metadata import Metadata
import matplotlib.pyplot as plt
import seaborn as sns
import torch
import random
import os.path
from os import listdir
from os.path import isfile, join
import faiss
import time
```

### Jeu de données PIMA

```python
# -----------------------------
# Load data
# -----------------------------
df = pd.read_csv(
    "Bureau/CHIMERA_SYNTHETIC_HEALTH_DATA_GENERATION/data/PIMA.csv"
)

# Metadata
metadata = Metadata.detect_from_dataframe(
    data=df,
    table_name="pima"
)

# -----------------------------
# Output directory
# -----------------------------
output_dir = (
    "Bureau/CHIMERA_SYNTHETIC_HEALTH_DATA_GENERATION/data/pima_synthetic_datasets_ctgan/"
)
os.makedirs(output_dir, exist_ok=True)

# -----------------------------
# Seeds
# -----------------------------
n_synth = 50
np.random.seed(123)
seeds = np.random.randint(1, 1000, size=n_synth)

times = []
synthetic_data_list = []

# -----------------------------
# Run CTGAN
# -----------------------------
for i in range(n_synth):

    start = time.time()

    synthesizer = CTGANSynthesizer(
        metadata,
        enforce_rounding=True,
        verbose=False
    )

    synthesizer.fit(df)

    synthetic_df = synthesizer.sample(num_rows=df.shape[0])

    elapsed = time.time() - start
    times.append(elapsed)

    synthetic_data_list.append(synthetic_df)

    # save dataset
    output_path = os.path.join(
        output_dir,
        f"PIMA_CTGAN_{i+1:02d}.csv"
    )
    synthetic_df.to_csv(output_path, index=False)

    print(f"Run {i+1}/{n_synth} - Time: {elapsed:.2f} sec")

# -----------------------------
# Runtime statistics
# -----------------------------
times = np.array(times)

mean_time = np.mean(times)
std_time = np.std(times)

execution_time_lower = mean_time - std_time
execution_time_upper = mean_time + std_time

print("\n===== CTGAN runtime summary =====")
print(f"Mean time: {mean_time:.2f} sec")
print(f"Std: {std_time:.2f} sec")
print(f"Interval: [{execution_time_lower:.2f}; {execution_time_upper:.2f}]")
```
